# Notebook for measuring average runtime from csv file

## Setup

```python

MEASURE="true_dtw_cy" # Either true_dtw_cy or true_frechet_cy
CITY="rome" # Either rome or porto
DATA_START_SIZE = 50
DATA_END_SIZE = 50
DATA_STEP_SIZE = 50
```

```python
import pandas as pd

csv_file_path = f'{CITY}/{CITY}_similarity_runtimes_{MEASURE}_start-{DATA_START_SIZE}_end-{DATA_END_SIZE}_step-{DATA_STEP_SIZE}.csv'

# Reading the CSV file
df = pd.read_csv(csv_file_path, index_col=0)

# Calculating the average for each column
averages = df.mean()

# Printing the average runtime for each dataset size
print("Average runtime for each dataset size:")
for dataset_size, avg_runtime in averages.items():
    print(f"{dataset_size}: {avg_runtime} ({round(avg_runtime, 3)})")
```
